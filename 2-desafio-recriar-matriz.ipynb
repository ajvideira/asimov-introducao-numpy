```python
import numpy as np
```

```python
#Solução inicial O(n*m)
matriz = np.empty((5,5), dtype=int)
for i in range(matriz.shape[0]):
    for j in range(matriz.shape[1]):
        if i == 0:
            matriz[i,j] = 5
        elif j == 0 or j == 4 or i == 4:
            matriz[i,j] = 3
        elif i == 2 and j == 2:
            matriz[i,j] = 15
        else:
            matriz[i,j] = 0
matriz
```

```python
#Solução otimizada com slicing
matriz = np.full((5,5), 3, dtype=int)
matriz_intermediaria = np.full((3,3), 0, dtype=int)
matriz_intermediaria[1,1] = 15
matriz[1:-1, 1:-1] = matriz_intermediaria
matriz[0,:] = 5
matriz
```
